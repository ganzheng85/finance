```python
def fetch_eod_data(symbol, from_date, to_date):
    """Fetches EOD data using EODHD API and returns a pandas DataFrame."""
    try:
        data = api.get_eod_historical_stock_market_data(
            symbol=symbol,
            from_date=from_date,
            to_date=to_date,
            order='a' # 'a' for ascending date order
        )
        # The library automatically processes the JSON into a structured format
        # and converts to a DataFrame
        df = pd.DataFrame(data)
        df['date'] = pd.to_datetime(df['date'])
        df = df.set_index('date').sort_index()
        return df
    except Exception as e:
        print(f"Error fetching data for {symbol}: {e}")
        return pd.DataFrame()

# Example Usage for Apple stock:
aapl_prices = fetch_eod_data(
    symbol='AAPL', 
    from_date='1995-01-01', 
    to_date='2025-12-31'
)

print(aapl_prices.head())
```

```python
# read universal ticker list 
try:
    universe_ticker_price = pd.read_csv("universe_ticker_price.csv")
    last_date = universe_ticker_price['date'].max()
    print(last_date)
except Exception as e:
    print(f"Error reading data: {e}")
    
universe_tickers = pd.read_csv("universe_tickers.csv")
universe_ticker_list = universe_tickers['ticker'].drop_duplicates().to_list()
print(len(universe_ticker_list))

universe_ticker_list  = [x for x in universe_ticker_list if isinstance(x, str)]

print(len(universe_ticker_list))
```

```python
last_date = '2026-01-02'
```

```python
import pandas as pd
from datetime import datetime

# List of tickers to iterate through
# ticker_list = ['AAPL', 'MSFT', 'INVALID_TKR', 'GOOGL']
ticker_list = universe_ticker_list + ['SPY', 'QQQ']
# ticker_list = ['SPY', 'QQQ']
# from_date = '1995-01-01'
# to_date = '2026-01-04'
try:
    from_date = last_date + pd.DateOffset(days=1)
except:
    from_date = '1995-01-01'

to_date = datetime.now().strftime('%Y-%m-%d')


# Containers for results
success_dataframes = []
failed_tickers = []

print(f"Fetching data for {len(ticker_list)} tickers...")

for ticker in ticker_list:
    # Use your existing fetch function
    ticker = str(ticker) + '.us'
    try:
        df = fetch_eod_data(ticker, from_date, to_date)
        if not df.empty:
            # Reset the index to move 'date' into a column
            df = df.reset_index()
            
            # Add the 'symbol' column to create the long format
            df['symbol'] = ticker
            
            success_dataframes.append(df)
            print(f"Successfully fetched: {ticker}")
        else:
            print(f"No data found for: {ticker}")
            failed_tickers.append(ticker)
    except Exception as e:
        # Catch any unexpected errors (network, API limits, etc.)
        print(f"Unexpected error for {ticker}: {e}")
        failed_tickers.append(ticker)
        # The loop continues to the next ticker automatically
        continue

# Create the final Long Format table
if success_dataframes:
    # Concatenate all individual DataFrames vertically
    long_format_df = pd.concat(success_dataframes, ignore_index=True)
    
    # Reorder columns to put symbol first (optional)
    cols = ['symbol', 'date'] + [c for c in long_format_df.columns if c not in ['symbol', 'date']]
    long_format_df = long_format_df[cols]
    
    print(f"\nSuccessfully combined {len(success_dataframes)} tickers into long format.")
else:
    long_format_df = pd.DataFrame()
    print("\nNo data was successfully fetched.")

# Display results
print(f"Failed tickers: {failed_tickers}")
# print(long_format_df.head())
```
