# Magnificent 7 vs S&P 500: Stock Return Comparison

This notebook downloads daily price data for the Magnificent 7 stocks and the S&P 500, calculates normalized returns, and plots their performance for visual comparison.

## 1. Install and Import Required Libraries
We use yfinance to download stock data, pandas/numpy for data manipulation, and matplotlib for plotting.

```python
# If running in a fresh environment, uncomment below to install yfinance
# %pip install yfinance

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Download Historical Stock Data with yfinance
We will use the tickers for the Magnificent 7 and S&P 500 (^GSPC).

```python
# Define tickers for Magnificent 7 and S&P 500
magnificent7 = ['GOOG', 'MSFT', 'META', 'NVDA', 'AMZN', 'AAPL', 'TSLA']
magnificent7 = ['CMG', 'MCD', 'DPZ', 'Yum','TXRH']
sp500 = ['^GSPC']
tickers = magnificent7 + sp500
# tickers = ['GOOG']

# Download daily price data (last 5 years)
# start_date = '2025-12-10'
time_range = 10 * 365  # 5 years in days
start_date = (pd.Timestamp.today() - pd.DateOffset(days=time_range)).strftime('%Y-%m-%d')
end_date = pd.Timestamp.today().strftime('%Y-%m-%d')

data = yf.download(tickers, start=start_date, end=end_date)['Close']
data = data.dropna(how='all')
data.head()
```

```python
# data = yf.download(tickers, start=start_date, end=end_date)
```

## 3. Calculate Daily Returns for Each Stock
We will compute daily percentage returns for each ticker.

```python
# Calculate daily returns for each ticker
returns = data.pct_change().dropna()
returns.head()
```

## 4. Calculate Daily Returns for S&P 500
We will extract and plot the S&P 500 returns for comparison.

```python
# Extract S&P 500 returns
sp500_returns = returns['^GSPC']
sp500_returns.name = 'S&P 500'
sp500_returns.head()
```

## 5. Plot Normalized Returns for Magnificent 7 vs S&P 500

This chart shows the cumulative returns of each stock and the S&P 500, normalized to 1 at the start date for easy comparison.

```python
# Plot normalized cumulative returns and print final returns
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 7))

# Calculate cumulative returns, normalized to 1 at start
to_plot = returns.copy()
to_plot = (1 + to_plot).cumprod()
to_plot = to_plot / to_plot.iloc[0]

# Calculate final returns for sorting
final_returns = to_plot.iloc[-1].sort_values(ascending=False)

# Print final cumulative returns for each ticker
print('Final cumulative returns (normalized to 1 at start):')
for ticker, value in final_returns.items():
    print(f'{ticker}: {value:.2f}')

# Plot each ticker in order of final return
for ticker in final_returns.index:
    if ticker != '^GSPC':
        plt.plot(to_plot.index, to_plot[ticker], label=ticker)
# Plot S&P 500 last for emphasis
plt.plot(to_plot.index, to_plot['^GSPC'], label='S&P 500', linewidth=4, color='black')

# Print final cumulative returns for each ticker as percentage (return = final - 1)
print('Final cumulative returns (percentage, normalized to 1 at start):')
for ticker, value in final_returns.items():
    pct_return = (value - 1) * 100
    print(f'{ticker}: {pct_return:.2f}%')

# Add percentage return to legend labels
legend_labels = []
for ticker in final_returns.index:
    pct_return = (final_returns[ticker] - 1) * 100
    if ticker == '^GSPC':
        label = f"S&P 500 ({pct_return:.0f}%)"
    else:
        label = f"{ticker} ({pct_return:.0f}%)"
    legend_labels.append(label)

# Get handles and labels from the plot
handles, labels = plt.gca().get_legend_handles_labels()
# Map ticker (or 'S&P 500') to handle
label_map = {lbl: h for lbl, h in zip(labels, handles)}

# Reorder legend: all tickers except S&P 500, then S&P 500 last
sorted_labels = [legend_labels[i] for i, t in enumerate(final_returns.index) if t != '^GSPC']
if '^GSPC' in final_returns.index:
    sorted_labels.append([lbl for lbl in legend_labels if lbl.startswith('S&P 500')][0])
sorted_handles = [label_map[lbl.split(' ')[0] if not lbl.startswith('S&P 500') else 'S&P 500'] for lbl in sorted_labels]
plt.legend(sorted_handles, sorted_labels, loc='upper left')

plt.title('Normalized Cumulative Returns: Major Restaurants vs S&P 500')
plt.xlabel('Date')
plt.ylabel('Cumulative Return (Normalized)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# --- 1. DATA ACQUISITION & PREPARATION (MISSING PART) ---
# Define the list of tickers you want to compare
# Example: McDonald's, Yum! Brands, S&P 500
tickers = ['MCD', 'YUM', 'DRI', '^GSPC'] 
start_date = '2015-01-01'
end_date = '2024-12-31' # Adjust end date as needed

print(f"Fetching data for: {', '.join(tickers)} from {start_date} to {end_date}...")

try:
    # Download historical price data (Adjusted Close is best for returns)
    price_data = yf.download(tickers, start=start_date, end=end_date)['Adj Close']

    # Calculate daily percentage returns
    # The 'returns' DataFrame must be defined for your plotting code to work
    returns = price_data.pct_change().dropna()
    print(f"Data prepared. First return date: {returns.index.min().date}, Last return date: {returns.index.max().date}")

except Exception as e:
    print(f"An error occurred during data fetching/preparation: {e}")
    print("Please ensure your internet connection is active and the tickers are correct.")
    exit()

# --- 2. YOUR ORIGINAL PLOTTING AND REVIEWED LOGIC ---
plt.figure(figsize=(12, 7))

# Calculate cumulative returns, normalized to 1 at start
to_plot = returns.copy()
to_plot = (1 + to_plot).cumprod()
# Normalization step is not strictly needed after cumprod() if we start at 1, but harmless
to_plot = to_plot / to_plot.iloc[0] 

# Calculate final returns for sorting
# .iloc[-1] gets the last row (final cumulative returns)
# .sort_values(ascending=False) ensures the list is ordered
final_returns = to_plot.iloc[-1].sort_values(ascending=False)

# Print final cumulative returns for each ticker
print('\n--- Final Cumulative Returns (Normalized to 1 at start) ---')
for ticker, value in final_returns.items():
    print(f'{ticker}: {value:.2f}') # e.g., 2.50 means 150% return

# Plotting logic is correct, but there is a repeated plot instruction for the legend.
# We will use one plotting block and one consolidated legend block.

# Plot each ticker in order of final return
for ticker in final_returns.index:
    # Plot non-benchmark tickers with standard style
    if ticker != '^GSPC':
        plt.plot(to_plot.index, to_plot[ticker], label=ticker)

# Plot S&P 500 last for emphasis
# Ensure '^GSPC' exists in the DataFrame before plotting
if '^GSPC' in to_plot.columns:
    plt.plot(to_plot.index, to_plot['^GSPC'], label='S&P 500', linewidth=4, color='black')

# Add percentage return to the final legend labels
legend_handles = []
legend_labels = []

# Iterate through final_returns to build the ordered legend
for ticker in final_returns.index:
    value = final_returns[ticker]
    pct_return = (value - 1) * 100
    
    # Define the label based on the ticker
    if ticker == '^GSPC':
        label = f"S&P 500 ({pct_return:.0f}%)"
    else:
        label = f"{ticker} ({pct_return:.0f}%)"
        
    # Find the handle for the ticker we just plotted
    # Note: get_legend_handles_labels() must be called after all plot calls
    
    # Simple workaround: Just plot them and let Matplotlib handle the legend order
    # based on the order they were plotted (which is already sorted by return)
    # The previous plotting loop is already sorted by `final_returns.index`.
    
    # We will manually recreate the handles/labels for the final, clean plot
    # and use the simple labels used in the plot calls (`ticker` or `'S&P 500'`).
    
    legend_labels.append(label)
    
print('\n--- Final Cumulative Returns (Percentage) ---')
for ticker, value in final_returns.items():
    pct_return = (value - 1) * 100
    print(f'{ticker}: {pct_return:.2f}%')


# Get the handles and labels as currently plotted
handles, labels = plt.gca().get_legend_handles_labels()

# Reorder handles and apply the percentage-formatted labels based on the sorted `final_returns`
sorted_handles = []
sorted_labels_with_pct = []

# Map the ticker to its handle and the new formatted label
ticker_to_handle = dict(zip(labels, handles))

for ticker in final_returns.index:
    value = final_returns[ticker]
    pct_return = (value - 1) * 100
    
    if ticker == '^GSPC':
        label_text = f"S&P 500 ({pct_return:.0f}%)"
        plot_label = 'S&P 500' # The label used in the plot call
    else:
        label_text = f"{ticker} ({pct_return:.0f}%)"
        plot_label = ticker # The label used in the plot call

    if plot_label in ticker_to_handle:
        sorted_handles.append(ticker_to_handle[plot_label])
        sorted_labels_with_pct.append(label_text)

# Apply the new legend
plt.legend(sorted_handles, sorted_labels_with_pct, loc='upper left')

plt.title('Normalized Cumulative Returns: Tickers vs S&P 500 Benchmark')
plt.xlabel('Date')
plt.ylabel('Cumulative Return (Normalized to 1)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 6. S&P 500 Profit Margin and P/E Ratio (Past 50 Years)

This section visualizes the average annual profit margin and price-to-earnings (P/E) ratio for the S&P 500 over the last 50 years. Each data point represents the average value for a given year.

```python
import pandas as pd
import requests

# The specific URL that contains the S&P 500 P/E ratio table by year
url = 'https://www.multpl.com/s-p-500-pe-ratio/table/by-year'

print(f"Fetching S&P 500 P/E data from: {url}...")

try:
    # 1. Set a common User-Agent to avoid being blocked by the server
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
    }
    
    # 2. Fetch the HTML content
    response = requests.get(url, headers=headers)
    
    # 3. Use pandas to read all tables found on the page
    # It usually returns a list of DataFrames. We assume the data is the first table (index 0).
    tables = pd.read_html(response.text)
    
    if not tables:
        print("Error: No HTML tables were found on the page.")
    else:
        # Select the relevant table (usually the first one)
        pe_ratio_data = tables[0]
        
        # 4. Clean and process the DataFrame
        
        # Rename columns (based on the table structure: Date, Value)
        pe_ratio_data.columns = ['Date', 'PE_Ratio']
        
        # The 'Value' column sometimes contains characters like '†'. 
        # This regex replaces everything that is NOT a digit or a period.
        pe_ratio_data['PE_Ratio'] = pe_ratio_data['PE_Ratio'].astype(str).str.replace(r'[^\d.]', '', regex=True)
        
        # Convert the cleaned P/E Ratio column to a numeric type
        pe_ratio_data['PE_Ratio'] = pd.to_numeric(pe_ratio_data['PE_Ratio'], errors='coerce')
        
        # Drop any rows where the conversion to numeric failed
        pe_ratio_data = pe_ratio_data.dropna(subset=['PE_Ratio'])
        
        # 5. Display the final data
        print("\nSuccessfully extracted S&P 500 P/E Ratio Data:")
        print(pe_ratio_data.head(10)) 
        print(f"\nTotal rows extracted: {len(pe_ratio_data)}")
        
        # Uncomment the line below to save the data to a CSV file
        # pe_ratio_data.to_csv('sp500_pe_ratio_by_year.csv', index=False)

except requests.exceptions.RequestException as e:
    print(f"An error occurred while fetching the URL: {e}")
except ValueError as e:
    # Catches errors if read_html fails to find or parse tables
    print(f"Could not parse HTML tables: {e}")
```

```python
# Plot S&P 500 P/E Ratio by Year
df_plot = pe_ratio_data.copy().iloc
# Convert 'Date' to year (int)
df_plot['Year'] = df_plot['Date'].str.extract(r'(\d{4})').astype(int)
df_plot = df_plot.sort_values('Year')

plt.figure(figsize=(12,6))
plt.plot(df_plot['Year'], df_plot['PE_Ratio'], marker='o', color='tab:red')
plt.title('S&P 500 P/E Ratio by Year')
plt.xlabel('Year')
plt.ylabel('P/E Ratio')
plt.grid(True)
plt.tight_layout()
plt.show()
```
